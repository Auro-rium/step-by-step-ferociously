```python
from typing import TypedDict , List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):

    values : List[str]
    name : str 
    result : str    
```

```python
def process_value(state: AgentState) -> AgentState:
    """ This function handles multiple diffrent inputs"""
    print(state)
    state["result"] = f"Hi there {state["name"]}! Your sum ={sum(state["values"])}"
    print(state)
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("proccessor", process_value)

graph.set_entry_point("proccessor")
graph.set_finish_point("proccessor")

app = graph.compile()
```

```python
from IPython.display import Image , display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
answers = app.invoke({"values":[1,2,8,4],"name":"Ishan"})
```

```python
print(answers["result"])
```
